```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Price (₹)': [200, 250, 300, 350, 400],
    'Fish Sold': [120, 100, 80, 55, 40]
}
df = pd.DataFrame(data)

print("="*70)
print("🐠 ANGLFISH PRICE TESTING - SALES DATA")
print("="*70)
print("\n")

print("ORIGINAL DATASET:")
print(df)
print("\n")
```

```python
df['Revenue (₹)'] = df['Price (₹)'] * df['Fish Sold']

print("UPDATED DATASET WITH REVENUE:")
print(df)
print("\n")
```

```python
max_revenue = df['Revenue (₹)'].max()
best_price = df[df['Revenue (₹)'] == max_revenue]['Price (₹)'].values[0]
best_quantity = df[df['Revenue (₹)'] == max_revenue]['Fish Sold'].values[0]

print("="*70)
print("📊 REVENUE ANALYSIS RESULTS")
print("="*70)
print(f"🏆 HIGHEST REVENUE: ₹{max_revenue:,}")
print(f"💰 BEST PRICE FOR ANGLFISH: ₹{best_price}")
print(f"📦 FISH SOLD AT BEST PRICE: {best_quantity} fish")
print("\n")
```

```python
plt.figure(figsize=(10, 6))
bars = plt.bar(df['Price (₹)'], df['Revenue (₹)'], 
               color=['#0077BE', '#0099CC', '#FF6B35', '#48CAE4', '#90E0EF'],
               edgecolor='#023E8A',
               linewidth=2)

for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height + 1000,
             f'₹{int(height):,}',
             ha='center', va='bottom', fontsize=11, fontweight='bold')

for bar in bars:
    if bar.get_height() == max_revenue:
        bar.set_color('#FF6B35')
        bar.set_edgecolor('#D62828')
        bar.set_linewidth(4)
        plt.text(bar.get_x() + bar.get_width()/2., height + 3000,
                 ' BEST PRICE ',
                 ha='center', va='bottom',
                 fontsize=12, fontweight='bold', color='#D62828')

plt.title(' Angelfish: Price vs Revenue Analysis', fontsize=16, fontweight='bold')
plt.xlabel('Selling Price (₹)', fontsize=13, fontweight='bold')
plt.ylabel('Total Revenue (₹)', fontsize=13, fontweight='bold')
plt.xticks(df['Price (₹)'], fontsize=11)
plt.grid(axis='y', linestyle='--', alpha=0.3)
plt.gca().set_facecolor('#F0F8FF')
plt.tight_layout()
plt.show()
```
