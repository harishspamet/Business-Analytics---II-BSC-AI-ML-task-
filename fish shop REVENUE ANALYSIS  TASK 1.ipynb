```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Price': [200, 250, 300, 350, 400],
    'Fish Sold': [120, 100, 80, 55, 40]
}
df = pd.DataFrame(data)
df
```

```python
df['Revenue'] = df['Price'] * df['Fish Sold']

df

max_rev = df['Revenue'].max()
opt_price = df.loc[df['Revenue'].idxmax(), 'Price']

print(f"Highest Revenue: ₹{max_rev}")
print(f"Optimal Price: ₹{opt_price}")
```

```python
#Price vs Revenue  BARCHART
plt.figure(figsize=(8, 5))
plt.bar(df['Price'], df['Revenue'], color='steelblue', edgecolor='black')
plt.xlabel('Price (₹)')
plt.ylabel('Revenue (₹)')
plt.title('Angelfish: Price vs Revenue Analysis')
plt.xticks(df['Price'])
plt.tight_layout()
plt.show()
```
